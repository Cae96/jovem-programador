```python

def validar_nota(nome_aluno, numero_nota):
  """Validates the input for a student's grade."""
  while True:
    try:
      nota = float(input(f"Digite a {numero_nota} nota de {nome_aluno}: "))
      if 0 <= nota <= 10:
        return nota
      else:
        print("A nota deve ser entre 0 e 10.")
    except ValueError:
      print("Entrada inválida. Por favor, digite um número.")


quantidade_alunos = int(input("Quantos alunos são? "))

alunos = {}

for i in range(quantidade_alunos):
  while True:
    nome_aluno = input(f"Digite o nome do aluno {i+1}: ")
    if nome_aluno not in alunos:
      break
    else:
      print("Este nome já existe. Por favor, digite um nome diferente.")

  nota1 = validar_nota(nome_aluno, "primeira")
  nota2 = validar_nota(nome_aluno, "segunda")
  nota3 = validar_nota(nome_aluno, "terceira")

  alunos[nome_aluno] = [nota1, nota2, nota3]

print("\nNotas e médias dos alunos:")
for aluno, notas in alunos.items():
  media = sum(notas) / len(notas)
  print(f"{aluno}: Notas = {notas}, Média = {media:.2f}")
```
